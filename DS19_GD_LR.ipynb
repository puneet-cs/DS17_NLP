# 1. Batch Gradient Descent

```python
# Batch GD updates the weights or bias based upon complete data
```

```python
import numpy as np

x = np.array([1,2,3,4,5])  # feature : Number of year of Experience
y = np.array([5,6,7,8,9]) # Target Salary

def Gradient_descent(x,y):
    w_curr = b_curr = 10
    learning_rate = 0.1
    n = len(x)
    epoch = 20

    for i in range(epoch):
        y_pred = w_curr * x + b_curr  # 10 * [1,2,3,4,5] + 10 = [20, 30, 40, 50, 60]
        wd = (1/n) * np.sum((y_pred - y) * x)
        bd = (1/n) * np.sum((y_pred - y) * 1)
        w_new = w_curr - learning_rate * wd    
        b_new = b_curr - learning_rate * bd
        w_curr = w_new
        b_curr = b_new

        y_predicted = w_curr * x + b_curr   # updated y_predited value
        cost = (1/(2*n)) * sum([val**2 for val in (y_predicted - y)])
        print(f"EPOCH : {i}, W : , { w_curr }, B : {b_curr}, COST : {cost}")
        
Gradient_descent(x,y)
```

# Batch GD on Large dataset

```python
import numpy as np
import pandas as pd

df = pd.read_csv("income_dataset_2025.csv")

df
```

```python
x = df.Year
y = df.Income

# Exploding Gradient Problem: if input data values are large then Gradient cost start increasing

# scale : bring the data into a small fix range (-3 to 3)
# z_scaler = (x - x_mean) / x_std
x = (x - x.mean()) / x.std()
y = (y - y.mean()) / y.std()

def Gradient_descent(x,y):
    w_curr = b_curr = 0
    learning_rate = 0.01
    n = len(x)
    epoch = 100

    for i in range(epoch):
        y_pred = w_curr * x + b_curr  # 10 * [1,2,3,4,5] + 10 = [20, 30, 40, 50, 60]
        wd = (1/n) * np.sum((y_pred - y) * x)
        bd = (1/n) * np.sum((y_pred - y) * 1)
        w_new = w_curr - learning_rate * wd    
        b_new = b_curr - learning_rate * bd
        w_curr = w_new
        b_curr = b_new

        y_predicted = w_curr * x + b_curr   # updated y_predited value
        cost = (1/(2*n)) * sum([val**2 for val in (y_predicted - y)])
        print(f"EPOCH : {i}, W : , { w_curr }, B : {b_curr}, COST : {cost}")
        
Gradient_descent(x,y)
```
